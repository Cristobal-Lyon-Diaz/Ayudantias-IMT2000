## ¿Qué son las variables?

En Python, una variable es un contenedor que se utiliza para almacenar valores. Los valores pueden ser números, cadenas de texto, listas, diccionarios, entre otros tipos de datos.

## Declaración y asignación de variables

Para crear una variable en Python, es necesario declararla y asignarle un valor. Por ejemplo, para crear una variable llamada "edad" y asignarle el valor de 30, se puede hacer lo siguiente:


````python
year = 2026
month = 9
day = 21
````

En este caso, "year" es el nombre de la variable y 2026 es el valor asignado. La variable "month" ahora contiene el valor de 9, de igual forma se entiende para las otras variables.

## Tipos de datos

Python tiene varios tipos de datos que se pueden asignar a una variable, incluyendo:

- Números enteros (`int`)
- Números de punto flotante (`float`)
- Cadenas de texto (`str`)
- Booleanos (`bool`)
- Listas (`list`)
- Diccionarios (`dict`)

Para conocer el tipo de datos de una variable, se puede utilizar la función `type()`. Por ejemplo:

````python
nombre = "Agustina"
edad = 30
pi = 3.1416

print(type(nombre)) 
print(type(edad)) 
print(type(pi)) 
````


## Operaciones con variables

Las variables pueden ser utilizadas en operaciones matemáticas y lógicas. Por ejemplo:


````python
a = 10
b = 5

suma = a + b
resta = a - b
multiplicacion = a * b
division = a / b
modulo = a % b
potencia = a ** b

print(suma) # Output: 15
print(resta) # Output: 5
print(multiplicacion) # Output: 50
print(division) # Output: 2.0
print(modulo) # Output: 0
print(potencia) # Output: 100000
````

## Uso de variables string en Python

Las variables string son una de las formas más comunes de almacenar texto en Python. En este notebook, veremos cómo crear y manipular variables string en Python.
Creando variables string

Para crear una variable string en Python, simplemente asignamos un valor entre comillas a una variable:

````python
mensaje = "Hola, ¿cómo estás?"
mensaje = 'Hola, ¿cómo estás?'
````

## Concatenando variables string

Podemos concatenar dos o más variables string simplemente utilizando el operador +:

````python
nombre = "Juan"
apellido = "Pérez"
nombre_completo = nombre + " " + apellido

nombre_completo
````

En el ejemplo anterior, la variable ``nombre_completo`` tendrá el valor "Juan Pérez".
Accediendo a caracteres específicos

Podemos acceder a caracteres específicos en una variable string utilizando la sintaxis [n], donde n es el índice del carácter que queremos acceder. Es importante tener en cuenta que los índices en Python comienzan en 0.

Por ejemplo, si queremos acceder al primer carácter de una variable string, podemos hacer lo siguiente:

````python
mensaje = "Hola, ¿cómo estás?"
primer_caracter = mensaje[0]

primer_caracter
````

En este caso, la variable `primer_caracter` tendrá el valor "H".

También podemos acceder a caracteres contando hacia atrás desde el final de la cadena utilizando índices negativos. Por ejemplo, si queremos acceder al último carácter de una variable string, podemos hacer lo siguiente:

````python
mensaje = "Hola, ¿cómo estás?"
ultimo_caracter = mensaje[-1]

ultimo_caracter
````

En este caso, la variable ``ultimo_caracter`` tendrá el valor "?".
Obtener una subcadena

Podemos obtener una subcadena de una variable string utilizando la sintaxis [n:m], donde n es el índice del primer carácter que queremos incluir en la subcadena, y m es el índice del primer carácter que queremos excluir de la subcadena.

Por ejemplo, si queremos obtener la subcadena "cómo" de la variable ``mensaje``, podemos hacer lo siguiente:

````python
mensaje = "Hola, ¿cómo estás?"
subcadena = mensaje[7:11]

subcadena
````

# Loops ``for`` y ``while`` en Python

Los bucles ``for`` y ``while`` son estructuras de control de flujo que permiten repetir un bloque de código varias veces. En ``Python``, el bucle for se utiliza para iterar sobre una ``secuencia`` (como una ``lista`` o una ``tupla``) o una ``colección`` (como un ``diccionario`` o un ``conjunto``), mientras que el bucle ``while`` se utiliza para repetir un bloque de código mientras se cumple una determinada condición.

### El bucle ``for``

La sintaxis del bucle for en ``Python`` es la siguiente:

```py

for variable in secuencia:
    # Código a ejecutar

Donde ``variable`` es una variable que se asigna a cada elemento de la ``secuencia`` en cada ``iteración`` del bucle. El bloque de código que se encuentra indentado después de los dos puntos ``:`` se ejecutará una vez para cada elemento de la secuencia.

Por ejemplo, podemos utilizar un bucle ``for`` para ``iterar`` sobre una lista de números y mostrar cada número por pantalla:

````python
numbers = [1, 2, 3, 4, 5]
for number in numbers:
    print(number)
````

También podemos utilizar la función ``range()`` para generar una ``secuencia`` de números y utilizarla en un bucle ``for``. La sintaxis de la función ``range()`` es la siguiente:

```py

range([start], stop[, step])

Donde ``start`` es el número inicial de la secuencia (por defecto es ``0``), ``stop`` es el número final de la secuencia (no se incluye en la secuencia) y ``step`` es el tamaño del paso entre cada número (por defecto es 1). Por ejemplo, podemos utilizar un bucle ``for`` y la función ``range()`` para mostrar los números del 1 al 10:

````python
for i in range(1, 11):
    print(i)
````

el bucle ``for`` se puede utilizar para iterar sobre una variedad de ``secuencias``. Algunas de las ``secuencias`` comunes que se pueden utilizar con el bucle ``for`` son:

- ``Listas``: una ``lista`` es una colección ordenada y mutable de elementos. Se puede utilizar un bucle ``for`` para iterar sobre cada elemento de la ``lista`` en el orden en que aparecen. Por ejemplo:


````python
numbers = [1, 2, 3, 4, 5]
for number in numbers:
    print(number)
````

- ```Strings```: un ``string`` es una secuencia de caracteres. Se puede utilizar un bucle ``for`` para iterar sobre cada carácter de la cadena. Por ejemplo:

````python
word = "hello"
for letter in word:
    print(letter)
````

- ``Rangos``: un ``rango`` es una secuencia inmutable de números enteros. Se puede utilizar un bucle ``for`` para iterar sobre cada número en el rango. Por ejemplo:

````python
for i in range(1, 6):
    print(i)
````

### El bucle ``while``

La sintaxis del bucle ``while`` en ``Python`` es la siguiente:

```py

while condición:
    # Código a ejecutar

Donde ``condición`` es una expresión ``booleana`` que se evalúa ``antes`` de cada ``iteración`` del bucle. El bloque de código que se encuentra ``indentado`` después de los dos puntos ``:`` se ejecutará mientras la condición sea verdadera.

Por ejemplo, podemos utilizar un bucle ``while`` para mostrar los números del 1 al 10:

````python
i = 1
while i <= 10:
    print(i)
    i += 1
````

Es importante tener cuidado al utilizar un bucle ``while``, ya que si la condición nunca se cumple, el bucle se ejecutará indefinidamente y podría hacer que el programa se bloquee o que consuma una gran cantidad de recursos del sistema.

### La función ``len()``

La función ``len()`` es una función incorporada en ``Python`` que se utiliza para devolver la longitud (```lenght```) de una ``secuencia``, es decir, el número de elementos que contiene. La sintaxis de la función ``len()`` es la siguiente:

```python

len(secuencia)

Donde secuencia es la ``secuencia`` para la cual se desea obtener la longitud. La secuencia puede ser una ``lista``, una ``tupla``, una ``string``, un ``rango`` o cualquier otra secuencia.

Aquí hay algunos ejemplos de cómo usar la función ``len()`` en ``Python``:

````python
numbers = [1, 2, 3, 4, 5]
length = len(numbers)
print(length)  # Output: 5

word = "hello"
length = len(word)
print(length)  # Output: 5

t = (1, 2, 3, 4, 5)
length = len(t)
print(length)  # Output: 5

r = range(1, 6)
length = len(r)
print(length)  # Output: 5
````

En estos ejemplos, la función ``len()`` se utiliza para obtener la longitud de diferentes secuencias, como una ``lista``, una ``string``, una ``tupla`` y un ``rango``.

La función ``len()`` puede ser muy útil para iterar sobre una secuencia utilizando un bucle ``for``. Por ejemplo, si queremos imprimir cada elemento de una lista en una línea separada, podemos usar un bucle ``for`` junto con la función len() de esta manera:

````python
numbers = [1, 2, 3, 4, 5]
for i in range(len(numbers)):
    print(numbers[i])
````

Un int con el código de la credencial.

20

Invalido

Explicación
Las divisiones enteras empezando desde 20 son 20->6 (nos detenemos pues 6 tiene un solo dígito). Al sumar los últimos dígitos de cada número (0+6) nos da 6, que al no ser divisible por 4 nos dice que la credencial es inválida.

243

3
