#### Imprimir

Imprimir en Python, se refiere a mostrar en pantalla el valor de una variable, es una funcion fundamental en Python.

Por ejemplo, si quiero mostrar en pantalla la sexta potencia de 2, hago lo siguiente:

```python
# Asigno las variables
dos = 2
seis = 6

# Realizo la operación
sexta_potencia = dos ** seis

# Imprimo el resultado
print(sexta_potencia)
```

```python
# No es necesario tanta asignacio, como pueden ver, se puede calcular directamente
otra_forma = 2 ** 6
print(otra_forma)
```

```python
# Y pueden apreciar que son iguales
valor_de_verdad = bool(sexta_potencia == otra_forma)
print(valor_de_verdad)
```
