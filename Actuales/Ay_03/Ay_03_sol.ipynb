### **if - elif - else**

```python
"""
Reglas:
    (1) Debe  existir un if (siempre 1)
    (2) Puede haber varios elif (puede haber entre 0 e infinito)
    (3) Podría no haber else (puede haber 1 o ninguno)

"""

n = int(input())

# se asigna una condición booleana
condicion_1 = n > 0
condicion_2 = n == 0

if condicion_1:
    print("n es positivo")

elif condicion_2:
    print("n es 0")

else:
    print("n es negativo")
```

### **while**

```python
n = int(input())
i = 0

while i <= n:
    if i%2 == 0:
        print(f"{i} es menor a {n} y es par")
    i += 1
```

### **for**

```python
n = int(input())

for intento in range(n):
    print(f"Intento numero {intento + 1}")
```

```python
n = int(input())
lista_perros = list()

for cantidad in range(n):
    lista_perros.append(f"Perro {cantidad}")

print(lista_perros)
```

```python
for perro in lista_perros:
    print(f"El {perro} le ladro al vecino")
```

```python
for perro in range(len(lista_perros)):
    if perro % 2 == 0:
        print(f"Se adopto al {lista_perros[perro]} ya que era par")
        lista_perros[perro] = "PERRO ADOPTADO"
        

print(lista_perros)
```

```python
palabra = "Esternocleidomastoideo"
vocales = 0
lista_vocales = ["A", "E", "I", "O", "U", "a", "e", "i", "o", "u"]

for letra in palabra:
    for vocal in lista_vocales:
        if letra == vocal:
            vocales += 1

print(vocales)
```

## **Laboratorio 3**

Escribe un segundo programa donde **reciba un string**, cuente sus vocales y si el numero de vocales es impar, pida nuevamente un string

```python
vocales = 1
lista_vocales = ["A", "E", "I", "O", "U", "a", "e", "i", "o", "u"]

while vocales % 2 != 0:
    palabra = input()
    vocales = 0

    for letra in palabra:
        for vocal in lista_vocales:
            if letra == vocal:
                vocales += 1
                
```
