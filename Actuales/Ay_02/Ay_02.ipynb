### Indexación string

Un string es un tipo de variable indexable, es decir, se puede extraer un caracter en especifico o un pedazo de este si es necesario, debido a que sigue un orden creciente desde el 0.

```python
# Por ejemplo
texto = "Hola mundo! Tengo 3 perros y 4 gatos"
print(f"El string tiene {len(texto)} caracteres")
```

```python
indice_perros = texto.index("perros") - 2 # Buscamos el 3, por lo tanto a perros le restamos dos y nos debiese dar 3
print(texto[indice_perros])
```

```python
texto_2 = f"Charles Mariano Aranguiz tiene 2 copas americas y 1 sudamericana, usa el 20"
palmares = texto_2[texto_2.index("2"):texto_2.index(",")]
dorsal = texto_2[-2:]
print(f"Charles usa el {dorsal} y su palmares es: {palmares}")
```
