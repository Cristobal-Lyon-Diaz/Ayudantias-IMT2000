Introducción al Test de Hipótesis
1. Introducción

El Test de Hipótesis es una técnica estadística utilizada para tomar decisiones sobre las características de una población en base a la información contenida en una muestra. En esta clase, exploraremos los conceptos clave relacionados con el Test de Hipótesis y cómo aplicarlo en problemas de Ciencia de Datos.

2. Definiciones básicas

Comenzamos por establecer algunas definiciones básicas:

- Población: Es el conjunto completo de elementos que estamos estudiando y del cual queremos obtener conclusiones.
- Muestra: Es un subconjunto de la población utilizado para realizar inferencias sobre la misma.
-  Hipótesis nula (H0): Es una afirmación que establece que no hay diferencia o relación entre las variables en estudio.
-  Hipótesis alternativa (H1 o Ha): Es una afirmación que contradice la hipótesis nula y sugiere que hay una diferencia o relación entre las variables.

3. Pasos del Test de Hipótesis

El proceso general del Test de Hipótesis consta de los siguientes pasos:

a) Formulación de hipótesis: Definir las hipótesis nula y alternativa adecuadas para el problema en cuestión.

b) Selección del nivel de significancia: Determinar el nivel de confianza deseado para el test, generalmente representado por alpha (α). Es común utilizar α=0.05 o α=0.01.

El concepto de alpha, también conocido como nivel de significancia, es un valor crítico establecido antes de realizar un Test de Hipótesis. Representado por la letra griega α, es la probabilidad máxima que estamos dispuestos a aceptar de cometer un error tipo I al rechazar la hipótesis nula cuando en realidad es verdadera.

En otras palabras, alpha nos permite controlar el riesgo de tomar una decisión incorrecta al rechazar la hipótesis nula. Comúnmente, los valores de alpha más utilizados son 0.05 (5%) y 0.01 (1%), pero pueden variar según el contexto y la rigurosidad requerida.

c) Selección de la prueba estadística: Elegir una prueba estadística apropiada que se ajuste al tipo de datos y la pregunta de investigación.

d) Cálculo de la estadística de prueba: Realizar los cálculos necesarios para obtener el valor de la estadística de prueba.

La estadística de prueba es una medida calculada a partir de los datos de muestra en un Test de Hipótesis. Su objetivo es proporcionar una medida numérica que nos permita tomar decisiones sobre la validez de la hipótesis nula.

La elección de la estadística de prueba depende del tipo de prueba de hipótesis y del tipo de datos que se estén analizando. En general, se busca una estadística de prueba que tenga una distribución conocida bajo la hipótesis nula.

Algunos ejemplos de estadísticas de prueba comunes incluyen:

    Prueba t de Student: Se utiliza para comparar las medias de dos muestras o la media de una muestra con un valor teórico. La estadística de prueba en este caso es la diferencia entre las medias dividida por el error estándar de la diferencia.

    Prueba chi-cuadrado: Se utiliza para evaluar la independencia o la homogeneidad entre variables categóricas. La estadística de prueba en este caso es la suma de los residuos al cuadrado dividida por los valores esperados.

    Prueba F: Se utiliza para comparar la varianza de dos muestras o la varianza de una muestra con un valor teórico. La estadística de prueba en este caso es el cociente de las varianzas.

e) Cálculo del valor p: Calcular el valor p asociado a la estadística de prueba obtenida.

el valor p cuantifica la probabilidad de obtener los datos observados o datos más extremos si la hipótesis nula fuera cierta. Si el valor p es pequeño, generalmente menor que el nivel de significancia (alpha), se considera que hay evidencia suficiente para rechazar la hipótesis nula en favor de la hipótesis alternativa.

f) Toma de decisión: Comparar el valor p con el nivel de significancia establecido y tomar una decisión en base a esto.

g) Conclusiones: Interpretar los resultados del test y hacer conclusiones sobre las hipótesis planteadas.
4. Tipos de errores en el Test de Hipótesis

Durante el proceso de Test de Hipótesis, podemos cometer dos tipos de errores:

- Error tipo I: Rechazar la hipótesis nula cuando es verdadera.
- Error tipo II: No rechazar la hipótesis nula cuando es falsa.

Es importante comprender los conceptos de estos errores y cómo pueden afectar nuestras conclusiones.



Al interpretar el valor p, se siguen dos enfoques comunes:

- Si el valor p es menor que el nivel de significancia (alpha) predefinido, generalmente se rechaza la hipótesis nula. Esto indica que los datos proporcionan suficiente evidencia para respaldar la hipótesis alternativa.

- Si el valor p es mayor que el nivel de significancia (alpha), no se rechaza la hipótesis nula. Esto implica que no hay suficiente evidencia para respaldar la hipótesis alternativa, y los datos no proporcionan un fundamento sólido para rechazar la hipótesis nula.

Es importante destacar que el valor p no indica la probabilidad de que la hipótesis nula sea verdadera o falsa. Simplemente proporciona información sobre la fuerza de la evidencia en contra de la hipótesis nula en relación con los datos observados.

Genere mil muestras normales y calcule la cantidad de veces que se rechaza el test de shapiro wilk.

En resumen, el test de Shapiro-Wilk es utilizado para evaluar si una muestra de datos sigue una distribución normal. Al calcular la estadística de prueba y el valor p, podemos determinar si hay suficiente evidencia para rechazar la hipótesis nula de normalidad. Es importante destacar que el test de Shapiro-Wilk es sensible a desviaciones de la normalidad en la cola de la distribución.

Toma de decisión:

    Si el valor p es menor que el nivel de significancia (alfa) predefinido, se rechaza la hipótesis nula y se concluye que los datos no siguen una distribución normal.
    Si el valor p es mayor que alfa, no se rechaza la hipótesis nula y no hay suficiente evidencia para afirmar que los datos no siguen una distribución normal.

```python
import numpy as np
from scipy.stats import shapiro
import matplotlib.pyplot as plt


# Definir parámetros
n = 100  # Tamaño de la muestra
mu = 0  # Media
sigma = 1  # Desviación estándar
alfa = 0.05  # Nivel de significancia
```

```python
# Inicializar contador para contar las veces que se rechaza H0
rechazos = 0

# Realizar el test de Shapiro-Wilk mil veces
for i in range(1000):
    # Generar una muestra normal de tamaño n
    muestra = np.random.normal(mu, sigma, n)
    
    # Realizar el test de Shapiro-Wilk
    _, p_value = shapiro(muestra)
    
    # Verificar si se rechaza la hipótesis nula
    if p_value < alfa:
        rechazos += 1
```

```python
# Calcular la proporción de rechazos
proporcion_rechazos = rechazos / 1000

# Imprimir el resultado
print("Proporción de rechazos:", proporcion_rechazos)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import shapiro

# Definir parámetros
n = 1000  # Tamaño de la muestra
mu = 0  # Media
sigma = 1  # Desviación estándar
alfa = 0.05  # Nivel de significancia
num_muestras = 1000  # Número de muestras a generar
sample_size = 50  # Tamaño del sample a graficar

# Inicializar contador para contar las veces que se rechaza H0
rechazos = 0

# Realizar el test de Shapiro-Wilk y generar muestras
muestra_grafico = []
for i in range(num_muestras):
    # Generar una muestra normal de tamaño n
    muestra = np.random.normal(mu, sigma, n)
    
    # Realizar el test de Shapiro-Wilk
    _, p_value = shapiro(muestra)
    
    # Verificar si se rechaza la hipótesis nula
    if p_value < alfa:
        rechazos += 1
    
    # Agregar la muestra al sample para graficar
    if i < sample_size:
        muestra_grafico.append(muestra)

# Calcular la proporción de rechazos
proporcion_rechazos = rechazos / num_muestras

# Imprimir el resultado
print("Proporción de rechazos:", proporcion_rechazos)

# Graficar las muestras
plt.figure(figsize=(10, 6))
for muestra in muestra_grafico[0:100]:
    plt.plot(range(n), muestra, alpha=0.7)
plt.xlabel('Índice')
plt.ylabel('Valor')
plt.title('Sample de las muestras generadas')
plt.show()
```
