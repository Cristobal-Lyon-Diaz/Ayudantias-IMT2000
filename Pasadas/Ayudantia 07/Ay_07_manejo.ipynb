# DataFrame en Pandas

## ¿Qué es un DataFrame?

En la biblioteca Pandas, un DataFrame es una estructura de datos tabular bidimensional con columnas etiquetadas. Es similar a una hoja de cálculo o una tabla de base de datos, donde cada columna puede contener un tipo de datos diferente (por ejemplo, números, cadenas, fechas, etc.). Los DataFrames son extremadamente útiles para el análisis y manipulación de datos.

## Creación de columnas

Pandas proporciona una sintaxis sencilla y eficiente para crear nuevas columnas en un DataFrame. A continuación, se presentan algunos ejemplos comunes:

````python
# Crear una nueva columna 'Total' como suma de las columnas 'Columna1' y 'Columna2'
df['Total'] = df['Columna1'] + df['Columna2']

# Crear una nueva columna 'Longitud' que contiene la longitud de los valores de la columna 'Texto'
df['Longitud'] = df['Texto'].apply(len)

# Crear una nueva columna 'EsMayor' que contiene valores booleanos True si la columna 'Edad' es mayor que 18, y False en caso contrario
df['EsMayor'] = df['Edad'] > 18

# Crear una nueva columna 'Categoria' basada en los valores de la columna 'Puntaje'
df['Categoria'] = pd.cut(df['Puntaje'], bins=[0, 50, 100], labels=['Bajo', 'Alto'])

# Crear una nueva columna 'Clasificación' basada en los valores de la columna 'Puntaje'
df['Clasificación'] = df['Puntaje'].apply(lambda x: 'Bueno' if x >= 80 else 'Regular')

# Crear una nueva columna 'Clasificación' basada en los valores de la columna 'Puntaje'
df['Clasificación'] = df['Puntaje'].apply(lambda x: 'Bueno' if x >= 80 else 'Regular')
````

### Creación de columnas de manera iterativa

Pandas permite crear columnas en un DataFrame de manera iterativa utilizando bucles o iteraciones sobre las filas del DataFrame. A continuación, se presenta un ejemplo de cómo hacerlo:

````python
# Crear una nueva columna 'Resultado' utilizando una iteración sobre las filas del DataFrame
df['Resultado'] = None  # Inicializar la columna con valores nulos

for index, row in df.iterrows():
    # Obtener los valores necesarios para calcular el resultado
    valor1 = row['Columna1']
    valor2 = row['Columna2']
    
    # Realizar el cálculo o la operación deseada
    resultado = valor1 + valor2
    
    # Asignar el resultado a la columna 'Resultado' en la fila correspondiente
    df.at[index, 'Resultado'] = resultado
````

En el ejemplo anterior, se itera sobre cada fila del DataFrame utilizando el método iterrows(). Para cada fila, se obtienen los valores necesarios para realizar el cálculo o la operación deseada. Luego, se realiza el cálculo y se asigna el resultado a la columna 'Resultado' en la fila correspondiente utilizando el método at[]. Es importante inicializar la columna con valores nulos o vacíos antes de la iteración.
