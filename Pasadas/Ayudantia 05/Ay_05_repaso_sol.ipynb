## Import en Python

La palabra clave import se utiliza en Python para importar módulos y paquetes en un programa. Un módulo es un archivo que contiene definiciones y declaraciones de funciones, variables y clases, mientras que un paquete es una colección de módulos relacionados que se organizan en un directorio.
Importar un módulo

Para importar un módulo en Python, se utiliza la sintaxis import nombre_del_modulo. Por ejemplo, si tenemos un módulo llamado random.py que contiene una función llamada randint(a,b), podemos importarla de la siguiente manera:

```python
import random
```

```python
random.randint(1,9)
```

## Importar partes específicas de un módulo

También es posible importar sólo una parte específica de un módulo utilizando la sintaxis from import nombre_de_la_funcion. Por ejemplo, si sólo queremos importar la función randint(a,b) de random.py, podemos hacerlo de la siguiente manera:

```python
from random import randint
```

```python
randint(1,9)
```

## Renombrar un módulo o función importada

También es posible renombrar un módulo o función importada utilizando la sintaxis import nombre_del_modulo as nuevo_nombre o from nombre_del_modulo import nombre_de_la_funcion as nuevo_nombre. Por ejemplo, si queremos importar el módulo random.py y llamarlo mi, podemos hacerlo de la siguiente manera:

```python
import random as rm
```

```python
rm.randint(1,9)
```

Crear un juego de cachipun:

Menu:
 - 1 -> Jugar (vs la computadora)
 - 2 -> Salir
 
Jugar: 
 - te permite elegir entre piedra, papel o tijeras
 - simula la opcion de la computadora
 - luego verifica quien gana
 - el juego lo gana el mejor de 3

Salir:
 - cierra el programa

```python
from random import choice
```

```python
lista = ["2", 2, 4, ("hola", 4)]
choice(lista)
```

```python
def menu():
    # -------- funcion que tome la opcion del usuario
    decision = input("Quieres jugar (1) o salir (2):")
    decision = int(decision)
    
    return decision
    
```

Nota: El código siempre lee los if, los elif los lee en caso de no cumplirse el primer "if"...

Ejemplo:

    if condicion_1:
        paso_1
    
    elif condicion_2:
        paso_2
        
    if condicion_3:
        paso_3
    elif ... 
        
Si se cumple la condición_1, entra al paso_1, pero no al elif. Siempre entra a todos los if.

```python
def jugar():
    # ------- funcion que permite simular la partida
    
    victorias = 0
    derrotas = 0
    juego = True
    opciones = ["piedra", "papel", "tijera"]
    
    while juego:
        
        opcion = input("Escoje tu mano: ")
        opcion_pc = choice(opciones)
        
        print("Elegiste", opcion, ", el pc eligió:", opcion_pc)
        
        if (opcion == "papel" and opcion_pc == "tijera"):
            derrotas += 1
        
        elif (opcion == "papel" and opcion_pc == "piedra"):
            victorias += 1
            
        elif (opcion == "papel" and opcion_pc == "papel"):
            print("empate")
        
        elif opcion == "tijera":
            
            if opcion_pc == "piedra":
                derrotas += 1
            
            elif opcion_pc == "papel":
                victorias += 1
            
            else:
                print("empate")
        
        elif opcion == "piedra":
            
            if opcion_pc == "papel":
                derrotas += 1
            
            elif opcion_pc == "tijera":
                victorias += 1
            
            else:
                print("empate")
        
        if victorias == 2:
            juego = False
        
        elif derrotas == 2:
            juego = False
        
        else:
            print("siguiente ronda")
        
    
    if victorias == 2:
        print("Ganaste!")
    
    else:
        print("Perdiste")
        
        
        
```

```python
def salir():
    # ------- funcion que simula el salirse de la partida
    print("Vuelve pronto")
    
```

```python
def cachipun():
    decision = menu()
    
    if decision == 1:
        jugar()
    
    else:
        salir()
```

```python
cachipun()
```
