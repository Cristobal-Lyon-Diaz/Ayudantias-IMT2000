## Correlación de datos

La correlación de datos es una medida estadística que nos permite cuantificar la relación o asociación entre dos variables. Esta medida nos ayuda a comprender si existe una conexión entre las variables y si esta conexión es positiva, negativa o nula.

En Python, podemos utilizar la biblioteca pandas para calcular la correlación entre variables de manera sencilla.

Para calcular la correlación utilizando pandas, podemos utilizar el método ```.corr()``` aplicado a un DataFrame. Este método calculará la correlación. El coeficiente de correlación oscila entre -1 y 1, donde 1 indica una correlación positiva perfecta, -1 indica una correlación negativa perfecta y 0 indica una correlación nula.

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
````

````python
# Ejemplo 1: Correlación positiva perfecta
altura = np.arange(160, 200, 2)  # Generar 20 valores de altura entre 160 y 200 (incremento de 2)
peso = altura - 100

data_1 = pd.DataFrame({'Altura': altura, 'Peso': peso})

# Calcular la matriz de correlación
correlation_matrix_1 = data_1.corr()

# Mostrar la matriz de correlación
print("Matriz de correlación - Ejemplo 1:")
print(correlation_matrix_1)

# Generar gráfico de dispersión
plt.scatter(altura, peso)
plt.xlabel('Altura')
plt.ylabel('Peso')
plt.title('Correlación positiva perfecta')
plt.show()
````

En este ejemplo, hemos creado dos listas de datos: "altura" y "peso". La altura y el peso están positivamente correlacionados, lo que significa que a medida que la altura aumenta, el peso también aumenta en la misma proporción. El resultado muestra una matriz de correlación con un coeficiente de correlación de 1 y un gráfico de dispersión que ilustra la correlación positiva perfecta.

````python
# Ejemplo 2: Correlación negativa perfecta
altura = np.arange(160, 200, 2)  
peso = 240 - altura  

data_2 = pd.DataFrame({'Altura': altura, 'Peso': peso})

# Calcular la matriz de correlación
correlation_matrix_2 = data_2.corr()

# Mostrar la matriz de correlación
print("\nMatriz de correlación - Ejemplo 2:")
print(correlation_matrix_2)

# Generar gráfico de dispersión
plt.scatter(altura, peso)
plt.xlabel('Altura')
plt.ylabel('Peso')
plt.title('Correlación negativa perfecta')
plt.show()
````

En este ejemplo, hemos invertido la relación entre altura y peso. Ahora, la altura y el peso están negativamente correlacionados, lo que significa que a medida que la altura aumenta, el peso disminuye en la misma proporción. El resultado muestra una matriz de correlación con un coeficiente de correlación de -1 y un gráfico de dispersión que ilustra la correlación negativa perfecta.

````python
# Ejemplo 3: Correlación débil o nula
# Generar dos conjuntos de datos aleatorios
np.random.seed(42)
edad = np.random.randint(20, 60, 20)
ingresos = np.random.randint(50000, 80000, 20)

# Agregar ruido aleatorio a los ingresos
ingresos = ingresos + np.random.normal(0, 1000, 20)

# Crear el DataFrame
data = pd.DataFrame({'Edad': edad, 'Ingresos': ingresos})

# Calcular la matriz de correlación
correlation_matrix = data.corr()

# Mostrar la matriz de correlación
print("\nMatriz de correlación - Ejemplo de correlación nula:")
print(correlation_matrix)

# Generar gráfico de dispersión
plt.scatter(edad, ingresos)
plt.xlabel('Edad')
plt.ylabel('Ingresos')
plt.title('Correlación nula')
plt.show()
````

En este caso, el coeficiente de correlación entre "Horas de estudio" y "Rendimiento" es aproximadamente 0.116, lo cual indica una correlación muy débil o nula. Este valor cercano a cero sugiere que no existe una relación clara entre las horas de estudio y el rendimiento académico de los estudiantes en este conjunto de datos. En otras palabras, los cambios en las horas de estudio no están relacionados de manera significativa con los cambios en el rendimiento académico.

OJO!👀

En la ciencia de datos, es importante comprender la diferencia entre correlación y causalidad. Aunque estos dos términos están relacionados, NO SIGNIFICAN LO MISMO.

Es importante tener en cuenta que la correlación no implica causalidad. Solo porque dos variables estén correlacionadas, no significa que una variable cause directamente cambios en la otra. Las correlaciones pueden ser el resultado de una coincidencia, la influencia de una tercera variable o una relación espuria. Es necesario llevar a cabo un análisis adicional y considerar el contexto antes de inferir causalidad a partir de la correlación.
